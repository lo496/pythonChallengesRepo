Cleaning Data with Pandas Challenge

```python
import numpy as np
import pandas as pd

df = pd.read_json('books_dirty.json')

# 1. replacing values with wrong type and empty strings with NaN
for x in df.index:
    if type(df.loc[x, 'publication_year']) != int:
        df.loc[x, 'publication_year'] = np.nan


    if type(df.loc[x, 'page_count']) != int or df.loc[x, 'page_count'] < 1: # type: ignore
        df.loc[x, 'page_count'] = np.nan


    if type(df.loc[x, 'average_rating']) != float:
        if type(df.loc[x, 'average_rating']) == int:
            df.loc[x, 'average_rating'] = float(df.loc[x, 'average_rating']) # type: ignore
        else:
            df.loc[x, 'average_rating'] = np.nan


    if type(df.loc[x, 'ratings_count']) != int:
        df.loc[x, 'ratings_count'] = np.nan


    if type(df.loc[x, 'price_usd']) != float:
        if type(df.loc[x, 'price_usd']) == int:
            df.loc[x, 'price_usd'] = float(df.loc[x, 'price_usd']) # type: ignore
        else:    
            df.loc[x, 'price_usd'] = np.nan


    if type(df.loc[x, 'sales_millions']) != float:
        if type(df.loc[x, 'sales_millions']) == int:
            df.loc[x, 'sales_millions'] = float(df.loc[x, 'sales_millions']) # type: ignore
        else:
            df.loc[x, 'sales_millions'] = np.nan


    if type(df.loc[x, 'in_print']) != bool:
        if type(df.loc[x, 'in_print']) == str and df.loc[x, 'in_print'].lower().strip() == 'true': # type: ignore
            df.loc[x, 'in_print'] = True
        if type(df.loc[x, 'in_print']) == str and df.loc[x, 'in_print'].lower().strip() == 'false': # type: ignore
            df.loc[x, 'in_print'] = False
        else:
            df.loc[x, 'in_print'] = np.nan


    if type(df.loc[x, 'available']) != bool:
        if type(df.loc[x, 'available']) == str and df.loc[x, 'available'].lower().strip() == 'true': # type: ignore
            df.loc[x, 'available'] = True
        if type(df.loc[x, 'available']) == str and df.loc[x, 'available'].lower().strip() == 'false': # type: ignore
            df.loc[x, 'available'] = False
        else:
            df.loc[x, 'available'] = np.nan

    
    if type(df.loc[x, 'last_checkout']) != str or df.loc[x, 'last_checkout'] == "" or df.loc[x, 'last_checkout'].lower().strip() == "n/a": # type: ignore
        df.loc[x, 'last_checkout'] = np.nan

# 2. clamp publication year
df["publication_year"] = df["publication_year"].clip(1800,2026)

# 3. correct impossible values 
df["page_count"] = df["page_count"].clip(1,None)
df["average_rating"] = df["average_rating"].clip(1.0,5.0)
df["ratings_count"] = df["ratings_count"].clip(1,None)
df["price_usd"] = df["price_usd"].clip(0.0,None)

# 4. normalize dates in last_checkout
df['last_checkout'] = pd.to_datetime(df['last_checkout'], format='mixed')

# 5. correct capitalization in genres, languages, formats, and publishers
df["genre"] = df["genre"].str.title().str.strip()
df["language"] = df["language"].str.title().str.strip()
df["format"] = df["format"].str.title().str.strip()
df["publisher"] = df["publisher"].str.title().str.strip()

# 6. remove duplicates
df.drop_duplicates(subset = ['title','author'], inplace = True)

# 7. drop rows with empty fields
# df.dropna(inplace = True)
# all rows have empty fields, so I'll leave this commented out

print(df)
```
