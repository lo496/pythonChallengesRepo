Format Info Challenge

```python
def get_info():
    f_name = input("What is your first name?")
    l_name = input("What is your last name?")
    birthday = input("What is your birthday?")
    email = input("What is your email?")
    formatted_f_name = f_name.title()
    formatted_l_name = l_name.title()
    return f"{formatted_f_name}\n{formatted_l_name}\n{birthday}\n{email}"

print(get_info())
```

Format Username Challenge

```python
raw_username = " Py_Thon-ROCKS123!!! "
formatted_username = raw_username.lower()
formatted_username = formatted_username.replace(" ","").replace("!","").replace("-","_")
i: int = 0
contains_digit: str = "does not contain a digit"
while i < len(formatted_username):
    if formatted_username[i] in "0123456789":
        contains_digit = "contains a digit"
    i+=1
contains_python = "contains \'python\'" if "python" in formatted_username else "does not contain \'python\'"
first_8 = formatted_username[:8]
print(f"Cleaned username \'{formatted_username}\' {contains_digit}, {contains_python}, and has first 8 characters \'{first_8}\'")
```

Work with Lambdas Challenge

```python
students = [
    ('alice', 34), ('bob', 22), ('diana', 19)
]
students.sort(key = lambda student: student[1], reverse=True)
print(students)
```

Manipulate Tuples Challenge

```python
students = [
    ("Alice", 20, "A"),
    ("Bob", 19, "B"),
    ("Charlie", 21, "A"),
    ("Diana", 20, "C"),
]
print([s for s in students if s[2]=="A"])
print([s[0] for s in students])
students = [(s[0],s[1],"B") if s[1]==20 else s for s in students]
print(students)
```

Manipulate Sets Challenge
